#backpropagation
Input
  ↓
Forward pass
  ↓
Prediction
  ↓
Calculate error
  ↓
Backpropagation
  ↓
Calculate how each weight contributed to error
  ↓
Update weights
  ↓
Repeat

Forward propagation tells us the prediction.
Backpropagation tells us how to change the weights.

Mathematically, we calculate:
∂w/∂L
This is the gradient of the loss with respect to the weight.

Wnew = Wold−η∂w/∂L
η = learning rate

Multiply the local derivatives together as you travel backward.

#Problem of Learning & Risk Minimization

we calculate the average loss:
R=1/Ni=1∑L(yi,y^i)
This is called the empirical risk.

Loss → error for a particular prediction/example.
Risk → average loss over the training examples.

Backpropagation
calculates:
∂W/∂R
Then:
Gradient Descent
uses it:
#Wnew = Wold−η∂w/∂L

#gradient descent:

The gradient points toward the direction of increasing loss.
We want the opposite direction — toward decreasing loss.

#Regularization:
The model learned the training examples extremely well but doesn't generalize well to unseen data.

Regularization is a collection of techniques used to reduce overfitting.

With L2 regularization, we modify it:
Ltotal=L+λi∑wi2

#dropout

Dropout is another technique used to reduce overfitting.
During training, dropout randomly sets some neuron outputs to zero
It encourages the network to learn more distributed representations.

#batch normalisation
BatchNorm acts like a standardizer between layers:

        messy values
             ↓
        [200, 500, 900...]
             ↓
       Batch Normalization
             ↓
      [manageable values]
             ↓
        next layer

This can make optimization more stable and often easier/faster.

BatchNorm takes the values coming from a layer, normalizes them using batch statistics, and then learns how much to scale and shift them.
