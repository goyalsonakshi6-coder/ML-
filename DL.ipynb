An autoencoder is a neural network that learns to:
compress data → reconstruct the original data

#RNN , LSTM AND GRU

#RNN
These are neural networks designed for sequential data.
The key difference from the CNNs we've studied is:
CNNs are mainly designed to capture spatial patterns; RNN-family models are designed to use information from previous time steps.
Mathematically:
ht=f(Wxxt+Wht−1+b)

it is used where order matter or sequense cause for that we do need the prev memory help

A basic RNN may struggle to retain the information from far earlier in the sequence.

This is one of the main reasons LSTM was introduced

#LSTM

Previous memory
      ↓
 ┌──────────────┐
 │ Forget gate  │ → remove unnecessary information
 └──────────────┘
      ↓
 ┌──────────────┐
 │ Input gate   │ → add useful new information
 └──────────────┘
      ↓
 Updated memory
      ↓
 ┌──────────────┐
 │ Output gate  │ → produce hidden state
 └──────────────┘

Forget gate
ft=σ(Wf[ht−1,xt]+bf)

Input gate
it=σ(Wi[ht−1,xt]+bi)

Output gate:
ot=σ(Wo[ht−1,xt]+bo)

#GRU
Unlike LSTM, GRU doesn't have a separate cell state.
It mainly uses two gates:
Update gate
Decides how much previous information to keep versus new information.

Reset gate
Decides how much previous information to ignore when creating new information.


```python
import torch
import torch.nn as nn
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader

# dataset
transform = transforms.ToTensor()

train_data = torchvision.datasets.CIFAR10(
    "./data", train=True, download=True, transform=transform
)

test_data = torchvision.datasets.CIFAR10(
    "./data", train=False, download=True, transform=transform
)

train_data = torch.utils.data.Subset(train_data, range(5000))
test_data = torch.utils.data.Subset(test_data, range(1000))

train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64)


model = nn.Sequential(
    nn.Flatten(),
    nn.Linear(3072, 256),
    nn.ReLU(),
    nn.Linear(256, 128),
    nn.ReLU(),
    nn.Linear(128, 10)
)
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
for epoch in range(15):

    for images, labels in train_loader:

        outputs = model(images)

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print("Epoch:", epoch + 1, "Loss:", loss.item())

correct = 0
total = 0

model.eval()

with torch.no_grad():

    for images, labels in test_loader:

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total

print("MLP Accuracy:", accuracy, "%")
```

```python
import torch
import torch.nn as nn
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader

transform = transforms.ToTensor()

train_data = torchvision.datasets.CIFAR10(
    "./data", train=True, download=True, transform=transform
)

test_data = torchvision.datasets.CIFAR10(
    "./data", train=False, download=True, transform=transform
)

train_data = torch.utils.data.Subset(train_data, range(5000))
test_data = torch.utils.data.Subset(test_data, range(1000))
train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64)

model = nn.Sequential(
    nn.Conv2d(3, 32, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),

    nn.Conv2d(32, 64, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(2),

    nn.Flatten(),

    nn.Linear(64 * 8 * 8, 128),
    nn.ReLU(),

    nn.Linear(128, 10)
)

loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

for epoch in range(20):

    for images, labels in train_loader:

        outputs = model(images)

        loss = loss_fn(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print("Epoch:", epoch + 1, "Loss:", loss.item())

correct = 0
total = 0

model.eval()

with torch.no_grad():

    for images, labels in test_loader:

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total

print("CNN Accuracy:", accuracy, "%")
```
