tensor math and comparision operators

```python
x = torch.tensor([1,2,3])
y = torch.tensor([9,8,7])

z1 = torch.empty(3)
torch.add(x,y, out=z1)
print(z1)
z2 = torch.add(x,y)
print(z2)
z=x+y
print(z)

z = torch.true_divide(x,y)
```

```python
#inplace operations
t = torch.zeros(3)
t.add_(x)
#or
t +=x 
#t = t+x is not a inplace operation, it will create a copy first
print(t)

z = x.pow(2)
z = x**2 
print(z)
```

```python
#simple comparision
z=x>0
z=x<0
```

```python
#matrix multiplication
x1 = torch.rand((2,5))
x2 = torch.rand((5,3))
z = torch.dot(x,y)
x3 = x1.mm(x2)
```

```python
#matrix exponentation
matrix_exp = torch.rand(5,5)
print(matrix_exp.matrix_power(3))
```

```python
#element wise mult.
z=x*y
print(z)
```

```python
#dot product
z=torch.dot(x,y)
print(z)
```

```python
#batch matrix multiplication
batch=32
n=10
m=10
p=30

tensor1=torch.rand(batch , n , m)
tensor2=torch.rand(batch , m , p)
out_bmm = torch.bmm(tensor1 , tensor2) # (batch , n, p)
```

```python
#example of broadcasting
x1 = torch.rand((5,5))
x2 = torch.rand((1,5))
#x2 will copy or perform through all rows of x1
z= x1 - x2
z= x1 ** x2
```
